## Handle Exception Name With information
* tyr block in which exception occur
* Give Object To Error Class To Show Message

```python
num1 = int(input("Enter First Number"))

num2 = int(input("Enter Second Number"))
try:
    div = num1/num2
    print("Division Value Is",div)
except ZeroDivisionError as var:
    print(var)

print("This Code Will Be Run After Erroe Message")
```

## Which Class Error Occured
* If We Don't Know Which Class Error Occured Just Write Exception 
* print(obj.__class__) To Show The Class Name

```python
num1 = int(input("Enter First Number"))

num2 = int(input("Enter Second Number"))
try:
    div = num1/num2
    print("Division Value Is",div)
except Exception as var:
    print(var) # Show Error Message
    print(var.__class__) # Show The Class Erroe Name 

print("This Code Will Be Run After Erroe Message")
```
