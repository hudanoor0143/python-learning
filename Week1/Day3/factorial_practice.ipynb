## Practice Questoin
* Function OF Factorial n Numbers
* Handle Invalid Inputs

```python
class InvalidIntiger(Exception):
    pass
try:
    def factorial(n):
        if (n<=0):
            raise InvalidIntiger("Enter Only Positive Intiger")
        return n*factorial(n-1)
except Exception as var:
    print(var)  

print(factorial(int(input("Eneter A Number For Factoril: "))))
```
