## Practice Questoin
* Function OF Factorial n Numbers
* Handle Invalid Inputs

```python
class InvalidInteger(Exception):
    pass
def factorial(n):
    if n<0:
        raise InvalidInteger("Enter Only Positive Intiger")
    if n == 0: # THIS IS THE BASE CASE TO STOP RECURSIVE CALL
        return 1
    return n*factorial(n-1)
try:
    num = int(input("Eneter A Number For Factoril: "))
    print(factorial(num))

except InvalidInteger as var: 
    print(var)  

except ValueError: # IF USER ENTER A CHARACTER GIVE THIS ERROR
    print("Enter Only Integer Not A Character")   
```
