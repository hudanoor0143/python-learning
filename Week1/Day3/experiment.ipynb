## Find factorial of n parameter  without function

```python

n = 5
fac = 1
for i in range(1,n+1):   #n+1 is given bcz range exclude last value
    fac*= i     
print(fac)    
```

## Find factorial of n parameter  with function

```python

#It is effective if we want to take a factorial of different number we no need to write code again 
#we just call function with different arguments
def cal_fac(n):
    fac = 1
    for i in range(1,n+1):
        fac*=i
    print(fac)
cal_fac(5)  
cal_fac(10)      
```

## Return Even or Odd string 

```python
#return Even or Odd string 

def evn_odd(n):
    if n%2==0:
        print("This Number Is Even")
    else:
        print("Number Is Odd")  
#funcoin call with different arguments
evn_odd(5)    
evn_odd(6)      
```

```python
#show number 5 4 3 2 1 by recursion

def show(n):
    if n <= 0: # Base case
        return 1
    print(n)
    show(n-1) # n-1 will remain continue we have to give base case to stop
show(5)
```

## Find Factorial With Recursion

```python
def fac(n): 
    #base case
    if (n==0 or n==1):
        return 1
    else:
        return n*fac(n-1)
print(fac(5))    
```

## Calculate the Sun of n natural numbers by recursion 

```python
#calculate the sun of n natural numbers by recursion 
def sum(n):
    if n<=0:
        return 0
    else:
        return n+sum(n-1)
print(sum(10))    
```

```python
#Recursion print all element in a list by using index
def list_element(list,indx=0): #indx is defaul parameter
    if (indx == len(list)): #base case to stp recursive call
        return 0
    print(list[indx])    # print item by index 
    list_element(list,indx+1)  #recursive call ..increment in index
names = ["mehr","shizi","ammara","huda"]

list_element(names)
```
