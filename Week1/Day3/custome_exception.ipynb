## User Defined Exception 
* Create Exception Class From Base Exception Class
* Write raise Statement in try Block With Error Message
* Handle Error By Except Block

```python
# Write A Program For FiveDivisionError Exception
# 1: CREATE A EXCEPTION CLASS 
class FiveDivisionError(Exception):
    #this Clas Call When Five Division Error Occure
    pass 

try:
    num1 = int(input("Enter First NUmber: "))
    num2 = int(input("Enter Second Number: "))
    if num2==5:
        raise FiveDivisionError("Cannot Divided By Five")
    div = num1/num2
    print(div)

except (FiveDivisionError,ZeroDivisionError) as obj: # Two Error Are Given In One Block Built in and Custom
    print(obj) #This Line Print Error Meessage Which Error Is Occured
```
