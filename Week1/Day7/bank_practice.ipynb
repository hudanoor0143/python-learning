## inheritance in Bank Account (..e.g Saving and Checking)

```python
class bank:

    def __init__(self,owner,current_amount = 0):
        self.owner = owner
        self.current_amount = current_amount
        
    try:    
 
        def deposit(self,deposit_amount):
            self.deposit_amount = deposit_amount
            try:
                if self.deposit_amount <=0:
                    print("Enter correct balance")
                else:
                    self.current_amount+=self.deposit_amount
                    print(f"Your {self.deposit_amount} amount is deposit successfully ")
            except Exception as e:
                print("Enter only integer")        
            
    except Exception as e:
        print("enter correct balance")
    
    
    def withdraw(self,withdraw_amount):
        self.withdraw_amount = withdraw_amount

        if self.current_amount == 0:
            print("your current balance is 0 ")
        elif self.current_amount<self.withdraw_amount:
            print("Insufficient balance")    
        else:
            self.current_amount-=self.withdraw_amount
            print(f"you withdraw {self.withdraw_amount} Successfully")

            print(f"Now You Reamining Balance is {self.current_amount}")

    def check_balance(self):
        print(f"{self.owner} Accout Balance is {self.current_amount} ")

class Savingaccount(bank):
    def __init__(self,owner,current_amount=0,interest_rate=0.03):
        super().__init(owner,current_amount)
        self.interest_rate=interest_rate

    def add_interest(self):
        interest = self.check_balance*self.interest_rate
        self.check_balance+=interest
        print("interest added",interest ,"new balance",self.current_amount)

class Checkingaccount(bank):
    pass

obj = Checkingaccount("noor_ul_huda".capitalize())
obj.deposit(1000)
obj.withdraw(900)
obj.check_balance()
obj.withdraw(900)
```
