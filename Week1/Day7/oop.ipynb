## super() function
The super() function is used to access method of the parent class

```python
# Super() Function


class parent:
    def __init__(self,value):
        self.value = value
    def display(self):
        print(f"parent value : {self.value}")

class child(parent):
    def __init__(self,value,extra):
        super().__init__(value)
        self.extra= extra
    def display(self):
        super().display()
        print(f"child extra:{self.extra}")
child_object = child(10,90)
child_object.display()    
print(type(child_object))        
```

```python
# Hybrid Inheritance
class A:
    def method(self):
        print("This is parent class")

class B(A):
    def method(self):
        print("This is class B")

class C(A):
    def method(self):
        print("This is class C")

class D(B,C): # multiplr inheritance
    pass 
object = D()
object.method()                      
```

## private attribute and methods

private attributes and methods are meant to be used within the class and are not acessible from out side the class

##### in this example when we access the private method outside the class it will raise an erroe

```python
# private attribute and methods 

class Account:
    def __init__(self,name,pasw):
        self.name = name
        self.__pasward = pasw # this is private attribute 

obj = Account("noor","1234")
print(obj.name)
print(obj.pasward)    # we can not accecc this outside the class     
```

##### we access private attribute inside the function 

```python

# private attribute can access inside the class
class Account:
    def __init__(self,name,pasw):
        self.name = name
        self.__pasward = pasw # this is private attribute 

    def set_pas(self):
        print(self.__pasward)  # we can access inside the function this private attribute 


obj = Account("noor","1234")
print(obj.name)
# print(obj.pasward)
obj.set_pas()
```

## Add Two Complex Number

```python
class Complex:
    def __init__(self,real,img):
        self.real=real
        self.img= img

    def showNumber(self):
        print(self.real,"i +",self.img,"j")

    def __add__(self,obj):
        newReal = self.real+obj.real
        newImg = self.img+obj.img    
        return Complex(newReal,newImg)

num1 = Complex(1,2)
num1.showNumber()  

num2 = Complex(4,8)
num2.showNumber()

num3 = num1+num2
num3.showNumber()

```

## Example of Abstract Class with private Methods

```python
from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
    @abstractmethod
    def drive(self):
        pass

class Car(Vehicle):
    def __start_fuel_pump(self):
        print("fuel pump started....")
    def start_engine(self):
        self.__start_fuel_pump() # This private method can not be used outside the class
        print("Car Engine Started!..")

    def drive(self):
        print("Now You can drive The Car..")

class Motorcycle(Vehicle):
    def start_engine(self):
        print("MotorCycle starts!...")
    def drive(self) :
        print("Now you can drive the Bike") 
    def apply_break(self):
        print("Stop The Bike")  

car = Car()
car.start_engine()  
car.drive()
bike = Motorcycle()
bike.start_engine()
bike. drive()
bike.apply_break()            
```
