Sorted() can be used with any iterable,including Tuple,String or Set. And Always return a New List

```python
text ="string"
sort = sorted(text)
print(sort)
```

## sorted() Function Parameters
sorted(iterable,key= None,reverse=False)
* iterable: it can be any iterable 
* key: key can be a Function
* reverse: By defaul value of reverse is False (assending order).If True then sort in (decending order)

```python
# sorted in assending order
 
list =  [3,6,7,9,2,0]
sort = sorted(list)
print(sort)
```

```python
#  sorted in assending order
 
list =  [3,6,7,9,2,0]
sort = sorted(list,reverse= True)
print(sort)
```

```python
# sorting using key parameter
# sorted the words by there lenght
name = ["shanza","ammara","mehr","huda"]
sort = sorted(name,key = len)
print(sort)
```

```python
# sorting the list of Dictionaries
dic = [
    {"name":"mehr", "score":22},
    {"name":"ammara", "score":33},
    {"name":"shizi", "score":32},
    {"name":"huda", "score":33}
    ]
sort = sorted(dic,key = lambda x:x["score"])
print(sort)
```

```python
#sort the list of dictionary with define function
dic = [
    {"name":"mehr", "score":22},
    {"name":"ammara", "score":33},
    {"name":"shizi", "score":32},
    {"name":"huda", "score":33}
    ]

def get_score(list):
    return list["score"]

y = sorted(dic,key = get_score)
print(y)
```
