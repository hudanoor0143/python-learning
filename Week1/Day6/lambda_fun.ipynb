## lambda function
**syntax** : 
lambda arguments: expression

```python
print_name = lambda name : "Hello " + name
print(print_name("Noor"))
```

lambda function are mostly used with higher order functions
like map(),filter(),sorted()


```python
# lambda function with map() function
numbers = [1,2,3,4]

squared_numbers = list(map(lambda x:x**2,numbers))

print(squared_numbers)
```

```python
#lambda with the filter() Function
numbers = [8,6,5,3,12,55,32]
#filter out Even numbers from the list
squared_numbers = list(filter(lambda x:x%2==0,numbers))

print(squared_numbers)
```
