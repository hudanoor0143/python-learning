## filter() function

filter function return an iterator where the items are filtered through a function to test if the item is True or False
* syntax : filter(function,iterable)
* filter function return in object form- we have to convert it in tuple or list 

```python
# filter function return object form

letters = ["a","b","e","w","o","i","t","s"]
def filter_vowels(letters):
    vowels = ["a","e","i","o","u"]
    if letters in vowels:
        return True
    else:
        return False
filterd_vowels = filter(filter_vowels,letters) # return in object form
vowels = filterd_vowels
print(vowels)    
```

```python
#convert to tuple
 
letters = ["a","b","e","w","o","i","t","s"]
def filter_vowels(letters):
    vowels = ["a","e","i","o","u"]
    if letters in vowels:
        return True
    else:
        return False
filterd_vowels = tuple(filter(filter_vowels,letters)) # return in tuple

print(filterd_vowels) 
```

```python
x = [1,2,3,4,2,4,6,8,80,90]
y = list(filter(lambda x: x>2,x))
print(y)
```
