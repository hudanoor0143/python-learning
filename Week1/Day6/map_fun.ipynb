## map() functin 
* map() can take multiple iterable 
* syntax:  map(function ,iterable1,iterable2,...)
* map() return in object form so we have to convert it in list or tuple

```python
#map() return object form 
x = [1,2,3,4]

y = map(lambda n : n**2,x)
print(y)
```

```python
#map function with lambda function 
x = [1,2,3,4]
# list item increment by 1
y = list(map(lambda n : n+1,x))
print(y)
```

```python
x = ["1","2","3"]
# data transformation : string to Integer
y = list(map(int,x))
print(y)
```

## map() combine with filter()
filter function return the iterable of even numbers

map function apply this filtered iterable on lambda dunction

```python
numbers = [1,2,3,4,5,6]
y = list(map(lambda x :x**2,filter(lambda x:x%2==0,numbers)))
print(y)
```

## map() with the define function

```python
def add(x,y):
    return x+y

num1 = [1,3,5,6]
num2 = [3,6,7,8]
result = list(map(add,num1,num2))
print(result)
```
