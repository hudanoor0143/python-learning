## File Handling
##### File Handling refers to a process of performonig operation on file such as creating,reding,writing,append and closing a File

* **'r'**: Read mode (default). Opens the file for reading.
* **'w'**:  Write mode. Opens the file for writing, overwriting existing content or creating a new file.
* **'a'**:  Append mode. Opens the file for writing, adding new content to the end.
* **'x'**:  Exclusive creation mode. Creates a new file, but fails if the file already exists.
* **'b'**:  Binary mode. Used for non-text files.
* **'t'**:  Text mode (default). Used for text files.
* **'+'**:  Update mode. Allows both reading and writing

```python
file = open("myfile2.txt","r")
read = file.read()
print(read)
```

## open a file using with ... open() way

```python
#The .write() method in Python returns the number of characters written to the file

with open("myfile2.txt","r+") as file1:
   
    write_content = file1.write("this will erase all existing data \n and write the new characters")
    print(write_content)
```

```python
with open("myfile2.txt","r+") as file1:
   
    write_content = file1.write("this will erase all existing data \nand write the new characters")
    print(write_content) 

    file1.seek(0)
    content = file1.read()
    print("update content: ")
    print(content)
```

```python
with open("myfile2.txt","a+") as file1:
    new_content= file1.write("this content is add to the end of the above file.")

    file1.seek(0)
    content = file1.read()
    print(content)
```

```python
with open("myfile2.txt","r") as file1:
    content = file1.readlines()  # read all lines 
    print(content)
```

```python
with open("myfile2.txt","a+") as file1:
    
    
    print(file1.tell()) 
    content = file1.readline() # readline give only one line 
    print(content)

    #Since file1.tell() points to the end, and readline() reads from the current position to the next newline,
    #it finds nothing and returns an empty string
```

```python
with open("myfile2.txt","r") as file1:
    file1.seek(0)
    print(file1.tell())
    content = file1.read(9)
    print(content)
```

```python
with open("myfile2.txt","rb") as file1:
    file1.write("this is my new content tah ii wnt to append in the file...")
    file1.seek(-10,2)
    content = file1.read()
    print(content)
```

### Exclusive creation mode with "x" 
#### the "x" mode is only for one-time creation of a file.
#### give Error if the file exist

```python
with open("myfile2.txt","x") as file1:
    content = file1.read()
    print(content)

# give fileExistError    
```

```python
with open("myfile3.txt","x+") as file2:
    file2.write("Create a new file and add new content in this file ")
    file2.seek(0)
    content = file2.read()
    print(content)
```

#### copy one file data to another file

```python
with open("data_to_copy.txt","r") as f:
    data = f.read()

with open("data_to_copy2.txt","w") as f:
    f.writelines(data)
```
