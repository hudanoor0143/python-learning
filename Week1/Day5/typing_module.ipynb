## **typing Module**


The typing module in Python is used to support type hints, which help you specify the expected data types of variables, function arguments, and return values.

```python
import typing
print(dir(typing))
```

```python
# without typing module
def add(a,b):
    return a+b
add(1,2.2)
# You can't easily tell what type name should be.
```

```python
# With typing module 
from typing import Optional
def add(a:int,b:int) -> int:
    return a+b
add(1,22)

# Now it's clear: add takes two integers and returns an integer.
```

```python
# here we are giving list of integer and reults in integer
from typing import List
def sum_number(numbers:List[int]) -> int:
    return sum(numbers)
sum_number([1,2,3,4,1,1])
```

```python
# here we can not define what type of results we get 
def sum_number(n):
    return sum(n)
sum_number([1,2,3,4,0.1,1])
```

```python
# list of string and reult in integer 
from typing import List
def total_length(word: List[str]) -> int:
    return sum(len(word) for word in word)
total_length(["noor","shizi","mehr"])
```
