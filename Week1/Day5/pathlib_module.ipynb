## **pathlib Module**

```python
from pathlib import Path

print(Path.cwd()) #Current Workind Directory
```

**If we want to iterate files and folders in this current directory we use **iterdir()****

```python
from pathlib import Path

for p in Path().iterdir():
    print(p)
    
```

```python
from pathlib import Path

p = Path.cwd()

print(p.parent)

print()

print(p.parent.parent)
print()

print(p.parent.parent.parent)
print()
d = Path("..").absolute()
print(d)
e = Path("..").resolve()
print(e)
```

```python
# Create a file by using touct() method

from pathlib import Path 
p = Path.touch("temporary_file.txt")
for p in Path().iterdir():
    print(p)
```

```python
# rename file or folder
from pathlib import Path 
p = Path("temporary_file.txt")
p.rename("temp_file.txt")
for p in Path().iterdir():
    print(p)
```

```python
#Delete the file by using unlink
from pathlib import Path 
p = Path("temp_file.txt")
p.unlink()
print("temp_file.txt is deleted successfully")
```
