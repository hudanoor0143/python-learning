## **requests.get()**

function in Python's requests library is used to send HTTP GET requests to a specified URL. It retrieves data from a server and accepts parameters to customize the request.

synttax:  requests.get(url, params=None, **kwargs)

```python
import requests 
i = requests.get("https://github.com/hudanoor014")
print("Response with ERROR",i)  # it gives me response 404 error because of wrong user id 


r = requests.get("https://github.com/hudanoor0143")
print("Response without ERROR",r) # it gives me response 200 it means this is right url
# if this link is open in website it gives an error
print(r.url) # if we want to print url we have to use .url 
```

requests.put() : is used replaces an existing resource on the server with new data.
