## Collections Module

The collections module in Python provides specialized, high-performance container data types.

Containers are data structures **(like List, dict, set, tuple)** that can hold multiple elements

These are built-in and very useful — but sometimes they aren't enough for specific tasks

```python
import collections

# print(dir(collections))
print(*dir(collections),  sep="\n")
```

## **collections.Counter()**

This is used to Count how many times each item appears

it looks like a dict, but it is actually a Counter object — a specialized subclass of dict.

```python
from collections import Counter
print(*dir(Counter),sep ="\n")
```

```python
#Count items by using collection.Counter
from collections import Counter
c = Counter("banana") 
print(c)
d = Counter(cats=4, dogs=8)
print(d)
print(type(d))
## Counter() gives the result in decending order
item = Counter("hello")
print(item)
```

```python
#Counter with integer


from collections import Counter
x = [1,2,3,4,5,1,4,3,4,2,4,5,6,7,8,76,8,7]
y =Counter(x)
print(y)
```

```python
# Counter with strings


from collections import Counter
x = "helo how are you are you "
y = x.split()
print(y,end = "\n\n")
z = Counter(y)
print(z)
```

```python
# Count items By using regular dict 
items = "hello"
count = {}
for item in items:
    if item not in count:
        count[item] = 1
    else:
        count[item]+=1
print(count)
```

## **collections.defaultdict()**

It's used when you want a dictionary that automatically provides a default value for keys that don't exist

we have to give the default valu as the argument in defaultdict() to explain what type of default value you want 


default value can be **int, str, list, set**

## defaultdict(list)
defaultdct with **list** type

```python
# Deafult with the list
from collections import defaultdict
data = [("fruit","apple"),("fruit","banana"),("veg","carrot")]
grouped = defaultdict(list) # list as default value
for fruits,items in data:
    grouped[fruits].append(items) 
print(grouped)    

#for key,value in grouped
```

## add a list without using defaultdict

```python
d = {} # create empty dictionary
data = [("fruit","apple"),("fruit","banana"),("veg","carrot")] # items for key values
for fruit,item in data: #loop on data 
    if fruit not in d: # check if the dictionary is empty
        d[fruit] = [] # assign value to empty list 
        d[fruit].append(item) # now append values in list
    else:
        d[fruit].append(item)

print(d)
```

## defaultdict(int)
default dict with type **int**

```python
# Defaultdict with int
from collections import defaultdict
fruits = ["apple","banana","apple","banana","mango","apple","peach","apple"]
count = defaultdict(int)
for fruit in fruits:
    count[fruit]+=1 
print(count)    
```

## defaultdict(set)
defaultdict with type **set**

```python
from collections import defaultdict 
x =["noor","name","mehr","name","shizi"]
d = defaultdict( set )   
for i in x:
    d[i[0]].add(i) # here we making a set of "n" characters
print(d)   
       

```

## **defaultdict(str)**

```python
from collections import defaultdict

x = ["noor", "name", "mehr","mom", "name", "shizi"]
d = defaultdict(str)

for word in x:
    d[word[0]] += word 

# without quotation it adds 0th index word without space
print(d)
```

```python
from collections import defaultdict

x = ["noor", "name", "mehr","mom", "name", "shizi"]
d = defaultdict(str)

for word in x:
    d[word[0]]+=word+" " # add word in string with space

print(d)    
```

## **namedtuple()**

A namedtuple is like a lightweight, immutable class that lets you access tuple elements by name

```python
from collections import namedtuple

Book = namedtuple("book",["title","author","year"],defaults = ["unknown"])
print(Book)

book1 = Book("something","Noor",1999)
book2 = Book("anythig","ammara") # third argument use the defaults value 
print(f"{book1.title} is written by {book1.author},in {book1.year}")
print(f"{book2.title} is written by {book2.author},in {book2.year}")
```
