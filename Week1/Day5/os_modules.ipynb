## OS _ Module 

**Python Provide OS Module to Work With Different Directories**

**Impliment Different Methods Of os Module**

## os.getcwd()
We Can get the present working directory **(In thr form of string)** by using the getcwd() method of os module

```python
import os
print(os.getcwd())
```

## os.chdir()

We can change the current directory by using **chdir()** method 

The new path that we want to change into be must be supplied as a **String** to this method

We can use both the forward_slash **/** and back_slash **\\**

## What it does:

* it changes the current working directory of your oython script.

* It does no rename any directory or move any files.

```python
import os
os.chdir("C:")
print(os.getcwd())
with open("file.txt","w") as f:
    x = f.write("this is my file and i can just modify this file")
    print(x)
```

**Again move to current director in which we are**

```python
os.chdir("d:\python-learning\Week1")
print(os.getcwd())
```

## os.listdir()
 
All files and sub-directories inside a directory can be retrieved using the listdir() method

```python
import os
print(os.getcwd())
os.listdir()
# Fine sub-directory 
#os.listdir("Day4")
```

## os.mkdir()

In Python, we can make a new directory using the mkdir() method.

```python
print(os.getcwd())
os.mkdir("Day5")
```

## os.rename()

The rename() method can rename a directory or a file.

For renaming any directory or file, rename() takes in two basic arguments:
* the old name as the first argument
* the new name as the second argument.

```python
# Check the list of all dierctoris
os.listdir()
os.rename("Day5","Day6")
# Day5 directory name will be change into Day6
```

## os.rmdir() or os.remove()

we can use the remove() method or the rmdir() method to remove a file or directory.

```python
os.listdir()
os.remove("myfile.txt") # myfile.txt file will be removed
# os.rmdir("mufile.txt") # We can also use this method 
```

## os.path.exists()
Checks if a file or directory exists.

This method returns True if path exists otherwise returns False.

```python
print("F drive exists:  ",os.path.exists("f:/"))
print(" python-learning folder is exists in D drive:  ",os.path.exists("D:/python-learning"))
```

## os.name

Returns the name of the operating system module (e.g., 'posix', 'nt')

```python
import os
print(os.name)
if os.name == 'nt':
    print("Running on Windows")
elif os.name == 'posix':
    print("Running on Unix-like OS")
```
