```python
import time

start = time.time()

n = 100
fac = 1
for i in range(1,n+1):   #n+1 is given bcz range exclude last value
    fac*= i     
print(fac)

end = time.time()

print("Time taken:", end - start, "seconds")
```

```python
import time

start = time.time()

def fac(n): 
    #base case
    if (n==0 or n==1):
        return 1
    else:
        return n*fac(n-1)
print(fac(100)) 

end = time.time()

print("Time taken:", end - start, "seconds")
```
