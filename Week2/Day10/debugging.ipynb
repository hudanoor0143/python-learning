## assert Statement

#### assert statement has a condition or expression which is supposed to be always true. If the condition is false assert halts the program and gives an AssertionError.

## Example
### We will use assert statement to check the parameter and if the length of the passed list is zero thr program halts

```python
def avg(marks):
    assert len(marks)!=0
    return sum(marks)/len(marks)

mark1 = []
print("Average of mark1 is:",avg(mark1))
```

### Now let's pass another list which will satisfy the assert condition and see what will be our output.

```python
def avg(marks):
    assert len(marks) !=0 ,"List is empty"
    return sum(marks)/len(marks)

mark1 = [78,54,44,89,43,13]
print("average of mark1 is:",avg(mark1))

mark2 = []
print("average of mark2",avg(mark2))
```

```python
def inc(x):
    return x+1

def test_answer():
    assert inc(4) ==5 ,"its not right equation"
    return True

print(test_answer())    
```

### By using assert:- Execution stops immediately — it does not run the other two assertions

```python
def inc(x):
    return x+1

def test_answer():
    assert inc(4) ==3 ,"inc(4) function error"
    assert inc(0) ==1 ,"inc(0) error"
    assert inc(-1) ==2 ,"inc(-1) error"
    return True
print(test_answer())   
```

## Assertion about Expected Exception

```python
import pytest
def test_zreodivision():
    with pytest.raises(ZeroDivisionError):
        print("about to divide by zero")
        1/0
    print("ZerDivisionError successfully caught")    
    
print(test_zreodivision())
```
