## Regular Expression
**A Regular Expression (RegEx) is a sequence of characters that defines a search pattern**

**Python has a module named re to work with RegEx**

```python
import re 
pattern = "^a....s$" # start from a and end with s and between a and s must be 4 characters
test_string = "acroos"
result = re.match(pattern,test_string)
if result:
    print("successfull")
else:
    print("not successfull")    
#print(result)
```

## re.match() returns a Match object

```python
import re 
pattern = "^a....s$" # start from a and end with s and between a and s must be 4 characters
test_string = "acroos"
result = re.match(pattern,test_string)
print(result)
```

* The span (start and end position of match)

* The matched text

## To get the actual matched string, you need .group()

```python
import re 
pattern = "^a....s$" # start from a and end with s and between a and s must be 4 characters
test_string = "acroos"
result = re.match(pattern,test_string)
print(result.group())
```

## re.match() vs re.search()


* **re.match() only checks at the beginning of the string.**

* **re.search() looks anywhere in the string.**

```python
import re 
s = "A omputer science portal for you:"
match = re.search(r"portal",s)
print("Start Index: ",match.start())
print("End Index: ",match.end())
```

## re.findall()

### The re.findall() method returns a list of strings containing all matches.

```python

# Program to extract numbers from a string
import re
string = " hello 12 hi 18s , howday 34"
pattern = r"\d"
result = re.findall(pattern,string)
print(result)
```

* **\d → matches a single digit (0-9)**

* **+ → means "one or more"**

```python
# Program to extract numbers from a string
import re
string = " hello 12 hi 18s , howday 34"
pattern = r"\d+"
result = re.findall(pattern,string)
print(result)
```

```python
import re
str = "\n and \r are the escape sequence "
result = re.findall(r"[\n\r]",str)
print(result)
```

## re.split()

### The re.split method splits the string where there is a match and returns a list of strings where the splits have occurred.

```python
import re 
string = "split from integers like 1 3 44 and 90"
result = re.split(r"\d+",string)
print(result)
```

### You can pass maxsplit argument to the re.split() method. It's the maximum number of splits that will occur

```python
import re 
string = "split from integers like 1 3 44 and 90"
result = re.split(r"\d+",string,maxsplit =1)
print(result)
```

## re.sub()

### The method returns a string where matched occurrences are replaced with the content of replace variable.

```python
import re 
string = "abc 12\
    de 23 \n fd 45"
# matches all whitespaces chracters
# replace it with / slash
replace = "/"
new_string = re.sub(r"\s+",replace,string)
print(new_string)
```

### You can pass count as a fourth parameter to the re.sub() method.

```python
import re 
string = "abc 12\
    de 23 \n fd 45"
# matches all whitespaces chracters
# replace it with / slash
replace = "/"
new_string = re.sub(r"\s+",replace,string,count = 2)
print(new_string)
```

## re.subn()

### The re.subn() is similar to re.sub() except it returns a tuple of 2 items containing the new string and the number of substitutions made.

```python
import re 
string = "abc 12\
    de 23 \n fd 45"
# matches all whitespaces chracters
# replace it with / slash
replace = "/"
new_string = re.subn(r"\s+",replace,string)
print(new_string)
```

## re.search()

### Searches for first occurrence of character or pattern

```python
import re
string = "python is fun"
#check if python is at the begining
match = re.search(r"\Apython",string)
print(match) #containing an object
print(match.group()) # The group() method returns the part of the string where there is a match
```

```python
import re 
string = "8015 345 32 1"
# Tree digit number or Two digit number followed space
pattern = r"(\d{3}) (\d{2})"
match = re.search(pattern,string)
print(match.group(1))
print(match.group(2))
print(match.start()) # start index from 0 
print(match.end())
```

## re.compile()

### Regular expressions are compiled into pattern objects

```python
import re
string = "Aye, said Mr. Gibenson Stark"
pattern = re.compile("[a-e]") # return the list of character between a to e
print(pattern.findall(string))
```

```python
import re 
p = re.compile(r"\S+") # list of characters without spaces
print(p.findall("this is my code . 12"))
```

## example
### The code uses regular expressions to find and list word characters, sequences of word characters, and non-word characters in input strings.

```python
import re 
p = re.compile(r"\w")
print(p.findall("this is the # list * of words"))

p = re.compile(r"\w+")
print(p.findall("this is the # list * of words"))

p = re.compile(r"\W+")
print(p.findall("this is the # list * of words"))
```

## a. Extracting Names
#### Assume that every name starts with Mr or Ms or Mrs, with an optional dot, a space and then followed by alphanumeric characters

```python
import re 
p = re.compile(r"(Mr|Mrs|Ms)\.?\s+[A-Z][a-z]+(?:\s+[A-Z][a-z]+)?")
matches = p.finditer(content)

for match in matches:
    print(match.group(0))
    
```

## b. Extracting Date of Births
### Assume that the the date, month and year are of two, two and four digits respectively. Moreover, they are separated by either a dot, a hyphen or a slash

```python
import re 
p = re.compile(r"\d{2}\W\d{2}\W\d{4}")
matches = p.finditer(content)
for i in matches:
    print(i.group())
```

```python
import re 
p = re.compile(r"(\d{2})\W(\d{2})\W(\d{4})")
matches = p.finditer(content)
for i in matches:
    print("Date: ",i.group(1))
```

```python
import re 
p = re.compile(r"(\d{2})\W(\d{2})\W(\d{4})")
matches = p.finditer(content)
for i in matches:
    print("Month : ",i.group(2))
```

```python
import re 
p = re.compile(r"(\d{2})\W(\d{2})\W(\d{4})")
matches = p.finditer(content)
for i in matches:
    print("Year: ",i.group(3))
```

## c. Extracting Emails and Usernames

### **Valid Name Part:**

* Lowercase case alphabets
* Uppercase case alphabets
* Digits: 0123456789,
* dot: . (not first or last character)
* For simplicity assume no special characters allowed

### **Valid Domain Part:**

* Lowercase case alphabets
* Uppercase case alphabets
* Digits: 0123456789,
* Hyphen: - (not first or last character),
* Can contain IP address surrounded by square brackets: test@[192.168.2.4] or test@[IPv6:2018:db8::1].

```python
import re 
pattern = re.compile(r"[A-Za-z0-9_.+-]+@[A-Za-z0-9-]+\.[[a-z]{2,}")
mails = pattern.finditer(content)
for mail in mails:
    print(mail.group(0))
```

### if we just want to extrct user name

```python
import re 
pattern = re.compile(r"([A-Za-z0-9_.+-]+)@([A-Za-z0-9-]+)(\.[[a-z]{2,})")
mails = pattern.finditer(content)
for mail in mails:
    print(mail.group(1))
```

## d. Extracting valid Cell phones

#### Assume that every valid phone number consiste of 10 digits in three groups of three, three and four digits. The three groups are separated by either a -, . or / symbol

```python
import re 
pattern = re.compile(r"([0-9]{3})[.#-/]([0-9]{3})[.#-/]([0-9]{4})")
matches = pattern.finditer(content)
for match in matches:
    print(match.group())
```

```python
import re
pattern = re.compile(r"https?://(www.)?(\w+)(\.\w+)")
matches = pattern.finditer(content)
for match in matches:
    print(match.group())
```
