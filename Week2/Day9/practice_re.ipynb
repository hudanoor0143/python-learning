##  Emails
* Extract all valid email addresses (hint: ignore the robert[at]example[dot]com one).

* Match only .com email addresses.

```python
import re 
with open("data_file.txt",encoding = "utf-8") as file:
    content = file.read()
#all valid Emails
email_pattern = re.compile(r"([a-zA-Z0-9_.+%-]+)@([a-zA-Z0-9-]+)\.[a-zA-Z.]{2,}")

com_email_pattern = re.compile(r"([a-zA-Z0-9_.+]+)@([a-zA-Z0-9-]+)\.com") # only .com emails
all_emails = email_pattern.finditer(content)

print("All Valid Emails")

for match in all_emails:
    print(match.group())
print()
print("only .com Emails")
com_emails = com_email_pattern.finditer(content)
for match in com_emails:
    print(match.group())    
```

##  Phone Numbers

* Extract all phone numbers in any format:
 +1-800-555-0199, (555)555-5555, 123.456.7890, etc.

* Extract only international phone numbers (start with + or 00). 

```python
import re 
with open("data_file.txt",encoding = "utf-8") as file:
    content = file.read()
    

phone_number = re.compile(r"(\+?\d{1,3}[-.\s])?\(?\d{2,4}\)?[-.\s]?\d{3,4}[-.\s]?\d{4}")    

internetional_number = re.compile(r"(\+|00)\d{1,3}[-.\s]?\d{1,4}[-.\s]?\d{3,4}[-.\s]?\d{4}")

number =phone_number.finditer(content)
print("All Phone Numbers")
for match in number:
    print(match.group())
print()
number2 = internetional_number.finditer(content)
print("International Numbers Only")
for match in number2:
    print(match.group())    
```

## Names

* Extract all full names (between Name: and | Email:).


```python
import re 
with open("data_file.txt",encoding = "utf-8") as file:
    content = file.read()

full_name = re.compile(r"Name:\s(.+?)\s*\| Email") 

matches = full_name.finditer(content)
for match in matches:
    print("Name: ",match.group(1))  # by group(1) we only get the name 
```

## Dates

* Extract all dates regardless of format: YYYY-MM-DD, DD-MMM-YYYY, YYYY/MM/DD, etc. 

```python
import re 
with open("data_file.txt",encoding = "utf-8") as file:
    content = file.read()
date_pattern = re.compile(r"Date:\s*\d{2,4}[-/.](\d){2,4}[-./]\d{2,4}")

#date_pattern = re.compile(r"Date:\s*\d{2,4}[-/.]?[a-zA-Z0-9]{3}+[-./]\d{2,4}") #this give this format Date: 15-Mar-2024
matches = date_pattern.finditer(content)
for match in matches:
    print(match.group())    
```
